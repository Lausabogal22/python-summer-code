# Variables
Variables are containers of that hold specific information. They can hold information like numbers, letters, words, and sentences as well as true and false values. The information that the categories mentioned can fall under are called `data types`.

## Data Types
- Numbers:
  - Integers -> `int`
  - Floats -> `float`
- Text:
  - Strings -> `str`
- True/ False values:
  - Booleans -> `bool`
- Collections of variables:
  - Lists -> `list`
  - Dictionaries -> `dict`
  - Sets -> `set`

```python
# x is an integer
x = 10
# y is also an integer
y = 5
# z is a float
z = 10.5
# s is a string
s = "Hola Mundo!"
# prop is a boolean
cond = True
# l1 is a list of integers
l1 = list()
l1 = [1, 2, 3, 4, 5]
# l2 is a list of strings
l2 = list()
l2 = ["a", "b", "c", "d"]
# d is a dictionary of lists
d = dict()
d = {"1": l1, "2": l2}
# a is a set of animals (strings)
a = set()
a = {"dog", "cat", "penguin", "gnu"}

print(type(x))
print(type(y))
print(type(z))
print(type(s))
print(type(cond))
print(type(l1))
print(type(l2))
print(type(d))
print(type(a))

print("")

print(f"x + y = {x+y}\n")

print(f"x + y + z = {x+y+z}\n")

print(f"s = {s}\n")

print(f"prop = {cond}\n")

print(f"l1 = {l1}")
print(f"The value at index '0' in 'l1' = {l1[0]}")
print(f"The value at index '1' in 'l1' = {l1[1]}\n")

print(f"l2 = {l2}\n")

print(f"d = {d}")
print(f"The value at key '1' in 'd' = {d["1"]}\n")

print(f"a = {a}\n")
```

## Conditions
With these different variable data types, you can compare the values between them to see if they hold the same or different values.

- Equal to -> `==`
- Not equal to -> `!=`
- Less than -> `<`
- Less than or equal to -> `<=`
- Greater than -> `>`
- Greater than or equal to -> `>=`

```python
x = 10
y = 5
z = 10.5
print(f"Is variable x greater than variable y?:\n {x > y}\n")
print(f"Is variable x less than or equal to variable z?:\n {x <= z}\n")
```

## Indentation
An important aspect of Python syntax, or coding rules, is that Python uses *indentation* to show which lines of code belong inside which statements compared to other languages like **C** or **Java** which uses curly braces `{}`. Curly braces are reserved for dictionaries.

```python
cond = True
if cond:
  print("The condition is True!")
```

## Control Flow
With conditions, we can now dictate the behavior of a program to either accomplish something if `x` is true or false. We can also contiuously check if `x` is true or not with loops, which can execute code within the loop.

- Conditions:
  - `if`
  - `elif`
  - `else`

```python
val = 3
if val == 1:
  print("val = 1")
elif val == 2:
  print("val = 2")
else:
  print("val = 3")
```

## Functions
Functions group code to accomplish specific behavior or computations. Functions are defined with the `def` keyword, followed by the function name and parenthesis and finished with a colon `:` (eg. `def func():`). Within the parenthesis, you can specify a number of arguments which can be used inside the function. Functions can also include a return statement (`return x`), which can return a specific value wherever the function was called and saved to variable.

```python
#---- Begin function definition ----
def func(cond):
  if cond:
    return "Hello World!"
  else:
    return "Adios Mundo!"
#---- End function definition ----

en = True
result = func(en)
print(result)
```

## Viewing Output / Basic Debugging
Using the `print()` function, we can pass a string or variable to see it's contents through standard out, as well as format the output with `print(f"{x}")` to include both a string and a variable to have a more explicit output. Functions can also be used inside print statements like `type()`, for example, to view the type of a variable and print it to standard out.

```python
myString = "Hello World!"

print(myString)
print(f"myString contains '{myString}'")
print(type(myString))
```

## Libraries
Python allows the use of libraries, which are external python programs that allow us to use external functions from those libraries so we do not have to code the functionality from scratch ourselves. Using the `import` keyword, we can import those libraries, and even specify a sort of nickname. For example, `import numpy` or `import numpy as np`, where `numpy` is the external library we want to use and `as np` is us specifing the nickname we want to use. We can use any number of libraries we want to as long as it is specified at the top of our python file. In ML workshop, y'all will be using `pandas`, `numpy`, and `matplotlib`.

### pandas
`pandas` is an open-source data analysis and manipulation tool built on top of python. It allows us to read and write data to different formats like CSV, text files, Excel, and SQL Databases. It also handles data alignment and missing data. `pandas` can also slice, index, and subset data.

### numpy
`NumPy` is an open-source numerical computing library for use in python. It allows for all sorts of scientific computations as well as adds mathematical functions to python to allow for intuitive calculations. It also allows support for large multi-dimensional arrays and matrices.

### matplotlib
`Matplotlib` is an open source library used to create static, dynamic, or interactive visualizations in Python. It allows for the creation of plots, graphs, and charts which can be generated through data provided to it.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

# Make charts look polished
plt.rcParams.update({
    "figure.facecolor": "white",
    "axes.facecolor":   "#F8FAFC",
    "axes.spines.top":  False,
    "axes.spines.right":False,
    "font.family":      "sans-serif",
})

# --- pandas example ---
print("--- pandas EXAMPLE START ---\n")
df = pd.DataFrame({'A': [1, 2, 3]})
print(f"pandas DataFrame:\n{df}\n")

# --- numpy example ---
print("\n--- NumPy EXAMPLE START ---\n")
pi = np.pi
print(f"The value of pi provided by numpy is: {pi}\n")

# --- matplotlib ---
print("\n--- Matplotlib EXAMPLE START ---\n")
# create a figure containing a single axis
fig, ax = plt.subplots()
# plot some data on the axis
ax.plot([1, 2, 3, 4], [1, 4, 2, 3])
# show the figure
plt.show()
```

```python
# Task:
# define 2 variables for customers that are of the type 'dictionary' ({"key": "value"})
# the dictionary should hold:
#   - the customer's name (string)
#   - how much they spent (int)
#   - and their age (int)
# --- WRITE YOUR CODE BELOW THIS LINE ---
customer_1 = {"name": "David", "spent": 1200, "age": 35}
customer_2 = {"name": "Richard", "spent": 100, "age": 65}

# --- begin predict_customer_tier definition ---
# predict_customer_tier should take in one argument for the customer data (dictionary variables we defined above)
# the function will check the customer's spent "key"
# if 'spend' is more than 1000, return "VIP Tier"
# else if the 'spend' value is more than 500, return "Standard Tier"
# else, return "Basic Tier" ('spend' is less than 500)
# --- WRITE YOUR CODE BELOW THIS LINE ---
def predict_customer_tier(customer_data):
  spend = customer_data["spent"]

  if spend > 1000:
    return "VIP Tier"
  elif spend > 500:
    return "Standard Tier"
  else:
    return "Basic Tier"

# --- end predict_customer_tier definition ---

# --- output ---
# Task:
# create a variable for a customer name from the first customer we defined
# and "index" their name (eg. customer_1["name"])
# --- WRITE YOUR CODE BELOW THIS LINE ---
name = customer_1["name"]
print(f"Predicting the tier of customer: {name}")

result = predict_customer_tier(customer_1)

print(f"{name}'s tier is: {result}!")
```
